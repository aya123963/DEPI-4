```python
"""
Idea:
- answer[i] = product of all elements to the left * product of all elements to the right

We do it in 2 passes:
1) Left pass → store prefix product in answer
2) Right pass → multiply by suffix product on the fly

No division used.
"""
```

```python
def productExceptSelf(nums):
    n = len(nums)
    answer = [1] * n

    # Step 1: prefix products (left side)
    prefix = 1
    for i in range(n):
        answer[i] = prefix
        prefix *= nums[i]

    # Step 2: suffix products (right side)
    suffix = 1
    for i in range(n - 1, -1, -1):
        answer[i] *= suffix
        suffix *= nums[i]

    return answer
```

```python
nums = [1,2,3,4]
print(productExceptSelf(nums))
```
